```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from scipy.signal import convolve2d

# -----------------------------------------
# 1. Read the image
# -----------------------------------------
img = io.imread('/content/image1.avif', as_gray=True)

# -----------------------------------------
# 2. Define 3x3 Laplacian filter mask
# -----------------------------------------
laplacian_mask = np.array([
    [0,  1,  0],
    [1, -4,  1],
    [0,  1,  0]
])

# -----------------------------------------
# 3. Apply Laplacian filter
# -----------------------------------------
filtered = convolve2d(
    img,
    laplacian_mask,
    mode='same',
    boundary='symm'
)

# -----------------------------------------
# 4. Normalize result for display
# -----------------------------------------
filtered_display = np.abs(filtered)

filtered_display = np.clip(
    filtered_display * 3,
    0,
    1
)

# -----------------------------------------
# 5. Display original image
# -----------------------------------------
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# -----------------------------------------
# 6. Display Laplacian filtered image
# -----------------------------------------
plt.subplot(1, 2, 2)
plt.imshow(filtered_display, cmap='gray')
plt.title('Laplacian Filtered Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

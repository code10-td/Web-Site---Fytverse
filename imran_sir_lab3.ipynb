```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from scipy.signal import convolve2d

# -----------------------------------------
# 1. Read the image
# -----------------------------------------
img = io.imread('/content/image1.avif', as_gray=True)

# -----------------------------------------
# 2. Define a 3x3 convolution mask
# -----------------------------------------
mask = np.array([
    [1/9, 1/9, 1/9],
    [1/9,  1/9, 1/9],
    [1/9, 1/9, 1/9]
])

# -----------------------------------------
# 3. Perform 2D convolution
# -----------------------------------------
result = convolve2d(img, mask, mode='same', boundary='symm')

# -----------------------------------------
# 4. Normalize result for display
# -----------------------------------------
result_display = np.abs(result)

# Normalize between 0 and 1
result_display = result_display / result_display.max()

# -----------------------------------------
# 5. Display images
# -----------------------------------------
plt.figure(figsize=(10, 4))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Convolution result
plt.subplot(1, 2, 2)
plt.imshow(result_display, cmap='gray')
plt.title('After 3x3 Convolution')
plt.axis('off')

plt.tight_layout()
plt.show()
```

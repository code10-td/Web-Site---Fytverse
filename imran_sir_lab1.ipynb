```python
import cv2
import matplotlib.pyplot as plt

# Read image in grayscale

img = cv2.imread('/content/speed-neon-lights-city.jpg'
# Check if image is loaded

if img is None:
    print("Image not found!")

else:
    # -----------------------------------
    # i) Histogram of Image
    # -----------------------------------
    histogram = cv2.calcHist(
        [img],
        [0],
        None,
        [256],
        [0, 256]
    )


    # -----------------------------------
    # ii) Low Pass Filtered Image
    # -----------------------------------
    low_pass = cv2.GaussianBlur(
        img,
        (5, 5),
        0
    )


    # -----------------------------------
    # iii) High Pass Filtered Image
    # -----------------------------------
    high_pass = cv2.Laplacian(
        img,
        cv2.CV_64F
    )

    # Convert the result to an unsigned 8-bit image
    high_pass = cv2.convertScaleAbs(high_pass)


    # -----------------------------------
    # Display all results
    # -----------------------------------
    plt.figure(figsize=(12, 8))


    # Original Image
    plt.subplot(2, 2, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")


    # Histogram
    plt.subplot(2, 2, 2)
    plt.plot(histogram)
    plt.title("Histogram of Image")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Number of Pixels")


    # Low Pass Filtered Image
    plt.subplot(2, 2, 3)
    plt.imshow(low_pass, cmap='gray')
    plt.title("Low Pass Filtered Image")
    plt.axis("off")


    # High Pass Filtered Image
    plt.subplot(2, 2, 4)
    plt.imshow(high_pass, cmap='gray')
    plt.title("High Pass Filtered Image")
    plt.axis("off")


    plt.tight_layout()
    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read original color image

img = cv2.imread('/content/speed-neon-lights-city.jpg')

# Check if image is loaded

if img is None:
    print("Image not found!")
    raise SystemExit

else:
    # Convert BGR to RGB for correct display in Matplotlib
    original_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # -----------------------------------
    # i) Histogram of Color Image
    # -----------------------------------
    plt.figure(figsize=(14, 10))


    # Original Image
    plt.subplot(2, 2, 1)
    plt.imshow(original_img)
    plt.title("Original Image")
    plt.axis("off")


    # Histogram
    plt.subplot(2, 2, 2)

    colors = ('b', 'g', 'r')

    for i, color in enumerate(colors):
        histogram = cv2.calcHist(
            [img],
            [i],
            None,
            [256],
            [0, 256]
        )

        plt.plot(histogram, color=color)

    plt.title("Histogram of Color Image")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Number of Pixels")


    # -----------------------------------
    # ii) Low Pass Filtered Image
    # -----------------------------------
    low_pass = cv2.GaussianBlur(
        original_img,
        (5, 5),
        0
    )

    plt.subplot(2, 2, 3)
    plt.imshow(low_pass)
    plt.title("Low Pass Filtered Image")
    plt.axis("off")


    # -----------------------------------
    # iii) High Pass Filtered Image
    # -----------------------------------
    high_pass = cv2.Laplacian(
        original_img,
        cv2.CV_64F
    )

    # Convert negative values to positive
    high_pass = cv2.convertScaleAbs(high_pass)

    plt.subplot(2, 2, 4)
    plt.imshow(high_pass)
    plt.title("High Pass Filtered Image")
    plt.axis("off")


    plt.tight_layout()
    plt.show()
```

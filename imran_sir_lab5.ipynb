```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from scipy.signal import convolve2d

# -----------------------------------------
# 1. Read the image
# -----------------------------------------
img = io.imread('/content/image1.avif', as_gray=True)

# -----------------------------------------
# 2. Define Horizontal Line Mask
# -----------------------------------------
horizontal_mask = np.array([
    [-1, -1, -1],
    [ 2,  2,  2],
    [-1, -1, -1]
])

# -----------------------------------------
# 3. Define Vertical Line Mask
# -----------------------------------------
vertical_mask = np.array([
    [-1,  2, -1],
    [-1,  2, -1],
    [-1,  2, -1]
])

# -----------------------------------------
# 4. Apply Horizontal Line Detection
# -----------------------------------------
horizontal = convolve2d(
    img,
    horizontal_mask,
    mode='same',
    boundary='symm'
)

# -----------------------------------------
# 5. Apply Vertical Line Detection
# -----------------------------------------
vertical = convolve2d(
    img,
    vertical_mask,
    mode='same',
    boundary='symm'
)

# -----------------------------------------
# 6. Convert results to positive values
# -----------------------------------------
horizontal = np.abs(horizontal)
vertical = np.abs(vertical)

# Normalize results
if horizontal.max() != 0:
    horizontal = horizontal / horizontal.max()

if vertical.max() != 0:
    vertical = vertical / vertical.max()

# -----------------------------------------
# 7. Display results
# -----------------------------------------
plt.figure(figsize=(15, 5))

# Original Image
plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Horizontal Lines
plt.subplot(1, 3, 2)
plt.imshow(horizontal, cmap='gray')
plt.title('Horizontal Lines')
plt.axis('off')

# Vertical Lines
plt.subplot(1, 3, 3)
plt.imshow(vertical, cmap='gray')
plt.title('Vertical Lines')
plt.axis('off')

plt.tight_layout()
plt.show()
```
